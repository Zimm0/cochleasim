Figuras diagnosticas/ilustrativas (sin comparacion contra referencia externa):

- Superficie |Pd(x,f)| en dB, para el punto 4 (analisis de estabilidad): muestra la cresta tonotopica y, aparte, la banda de resonancia de cavidad a baja frecuencia.

```python
import numpy as np
from cochleasim.models.params_loader import HUMAN_PARAMS, CAT_PARAMS
from cochleasim.analysis.response import compute_response_grid
from cochleasim.analysis.plots import plot_tuning_curves, plot_pd_surface
```

## Curvas de sintonia -- humano

## Superficie Pd(x,f) -- humano, 30 Hz-16 kHz

Ajusta `view=(elev, azim)` si el angulo no deja ver bien la banda de resonancia.

```python
plot_tuning_curves(HUMAN_PARAMS, [500, 1000, 2000, 4000, 8000]);
```

```python
freqs = np.logspace(np.log10(30), np.log10(16000), 120)
grid = compute_response_grid(HUMAN_PARAMS, freqs, N=1000)
plot_pd_surface(grid,view=(30, 60),max_freq_points=2000,max_position_points=2000);
```

## Superficie Pd(x,f) -- gato

```python
freqs_cat = np.logspace(np.log10(30), np.log10(65000), 120)
grid_cat = compute_response_grid(CAT_PARAMS, freqs_cat, N=1000)
plot_pd_surface(grid_cat,view=(30, 60),max_freq_points=2000,max_position_points=2000);
```
