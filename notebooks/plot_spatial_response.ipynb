```python
import matplotlib.pyplot as plt

from cochleasim.data_loaders import load_frequency_response_table, load_gerbil_cf_map
from cochleasim.models.spatial_reconstruction import (
    compute_spatial_responses,
    plot_spatial_responses,
)
```

```python
lookup_table = load_frequency_response_table()
cf_distance_cm, cf_hz = load_gerbil_cf_map()
```

```python
responses = compute_spatial_responses(
    lookup_table, cf_distance_cm, cf_hz,
    frequencies_hz=[10, 100, 500, 2000, 8000],
)
```

```python
fig, ax_amplitude, ax_phase = plot_spatial_responses(responses)
```
