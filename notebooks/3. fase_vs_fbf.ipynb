# Fase vs f/BF

Diagnostico sin referencia externa ni tabla de error: acumulacion de fase del modelo al acercarse a la CF de cada posicion, y sensibilidad al parametro H.

```python
import dataclasses
import numpy as np
from cochleasim.models.params_loader import HUMAN_PARAMS, CAT_PARAMS
from cochleasim.analysis.response import characteristic_place, tuning_curve_at_position, phase_vs_ratio
from cochleasim.analysis.plots import plot_phase_vs_ratio
```

## Fase vs f/BF -- varias CFs, humano y gato

```python
plot_phase_vs_ratio(HUMAN_PARAMS, [500, 1000, 2000, 4000, 8000, 11000]);
```

```python
plot_phase_vs_ratio(CAT_PARAMS, [500, 1000, 2000, 4000, 8000, 16000]);
```

## Comparacion cuantitativa: fase acumulada, humano vs. Tubelli et al.

Unica referencia externa real disponible para este diagnostico: acumulacion de fase entre f/BF=0.02 y f/BF=0.9, en un modelo de elementos finitos humano validado contra mediciones de hueso temporal.

```python
from cochleasim.analysis.response import phase_accumulation
from cochleasim.validation.reference import load_phase_accumulation_reference

delta = phase_accumulation(HUMAN_PARAMS, 11000, N=1000)
ref = load_phase_accumulation_reference('human')
print(f'modelo: {delta:.2f} ciclos  |  {ref.label}: {ref.y[0]:.2f} ciclos')
```
