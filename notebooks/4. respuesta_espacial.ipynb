# Respuesta espacial a frecuencia fija: xi_b(x) y Pd(x)

Envolvente de la onda viajera para varias frecuencias de estimulo fijas -- equivalente directo a las Fig. 5/6 del propio paper de Neely & Kim. xi_b se calibra en dB re: 1 nm usando Pe = 0 dB SPL (P_REF_DYN_CM2), igual que la convencion del paper.

```python
import numpy as np
from cochleasim.models.params_loader import HUMAN_PARAMS, CAT_PARAMS
from cochleasim.models.cochlear_model import P_REF_DYN_CM2
from cochleasim.analysis.response import compute_response_grid
from cochleasim.analysis.plots import plot_spatial_response
```

## Humano

```python
freqs = np.logspace(np.log10(100), np.log10(16000), 100)
grid = compute_response_grid(HUMAN_PARAMS, freqs, Pe=P_REF_DYN_CM2, N=1000)

plot_spatial_response(grid, freqs_to_show=[400, 1600, 6400, 2000, 4000, 8000], quantity='xi_b', ylim=(-100, 10));
```

```python
plot_spatial_response(grid, freqs_to_show=[250, 500, 1000, 2000, 4000, 8000], quantity='Pd', ylim=(-30, 60));
```

## Gato

```python
freqs_cat = np.logspace(np.log10(200), np.log10(30000), 100)
grid_cat = compute_response_grid(CAT_PARAMS, freqs_cat, Pe=P_REF_DYN_CM2, N=1000)

plot_spatial_response(grid_cat, freqs_to_show=[400, 1600, 6400, 25600], quantity='xi_b', ylim=(-100, 10));
```
